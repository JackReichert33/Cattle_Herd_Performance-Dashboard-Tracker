```python
#Sample Cattle Dataset

import pandas as pd

cattle = pd.DataFrame({
    "Animal_ID": [101, 102, 103, 104, 105, 106,
                  107, 108, 109, 110, 111, 112],
    "Breed": ["Angus", "Hereford", "Angus", "Hereford",
              "Angus", "Hereford", "Angus", "Hereford",
              "Angus", "Hereford", "Angus", "Hereford"],
    "Starting_Weight": [650, 700, 625, 675, 640, 720,
                        660, 690, 630, 710, 680, 645],
    "Ending_Weight": [950, 1020, 875, 975, 920, 1010,
                      960, 1000, 900, 1040, 980, 945],
    "Days_On_Feed": [100, 100, 100, 100, 100, 100,
                     100, 100, 100, 100, 100, 100],
    "Feed_Cost": [240, 255, 230, 248, 235, 260,
                  242, 250, 232, 265, 245, 238]
})

cattle.head()
```

```python
import pandas as pd

# Calculate total weight gain using the correct column names
cattle["Weight_Gain"] = (
    cattle["Ending_Weight"] - cattle["Starting_Weight"]
)

# Calculate Average Daily Gain (ADG) using Days_On_Feed
cattle["ADG"] = (
    cattle["Weight_Gain"] / cattle["Days_On_Feed"]
)

# Rank cattle from highest to lowest ADG
cattle["ADG_Rank"] = cattle["ADG"].rank(
    ascending=False, method="min"
)

# Label performance relative to your herd
average_adg = cattle["ADG"].mean()

cattle["Performance"] = cattle["ADG"].apply(
    lambda x: "Above Herd Average"
    if x > average_adg
    else "At or Below Herd Average"
)

# Display results with the existing columns
display(
    cattle[
        ["Animal_ID", "Breed", "Weight_Gain",
         "Days_On_Feed", "ADG", "ADG_Rank", "Performance"]
    ].sort_values("ADG_Rank").round(2)
)
```
